```python
import importlib.util
from importlib import metadata

# Các thư viện cần cài để train, tối ưu hóa, đánh giá model + tiền xử lý dữ liệu + hiển thị giao diện đẹp hơn
libraries = {
    "pandas": "Đọc/ghi và xử lý dữ liệu dạng bảng (csv, parquet, excel)",
    "numpy": "Tính toán mảng số học và xử lý dữ liệu hiệu năng cao",
    "seaborn": "Vẽ biểu đồ thống kê và khám phá dữ liệu (EDA)",
    "matplotlib": "Vẽ biểu đồ cơ bản và tùy chỉnh",
    "underthesea": "Tách từ, tách câu và xử lý ngôn ngữ tiếng Việt",
    "scikit-learn": "Tiền xử lý dữ liệu, metrics, train/test split",
    "torch": "Huấn luyện và suy luận mô hình Deep Learning",
    "transformers": "Sử dụng các mô hình Transformer như PhoBERT",
    "datasets": "Quản lý dataset của HuggingFace",
    "evaluate": "Đánh giá mô hình NLP",
    "accelerate": "Tăng tốc huấn luyện Transformer",
    "peft": "LoRA, QLoRA và fine-tuning tiết kiệm tài nguyên",
    "bitsandbytes": "Quantization 4bit/8bit",
    "pyarrow": "Đọc/ghi file parquet",
    "openpyxl": "Đọc/ghi file Excel (.xlsx)",
    "emoji": "Nhận diện và xử lý emoji",
    "wordcloud": "Vẽ word cloud cho EDA",
    "tqdm": "Thanh tiến trình",
    "jupyter": "Notebook để khám phá dữ liệu và thử nghiệm",
    "ipywidgets": "Widget tương tác trong Jupyter Notebook",
}

print("=" * 120)
print(f"{'THƯ VIỆN':<18} {'TRẠNG THÁI':<12} {'VERSION':<15} {'DÙNG ĐỂ'}")
print("=" * 120)

missing = []

for package, description in libraries.items():

    module_name = package

    if package == "scikit-learn":
        module_name = "sklearn"

    spec = importlib.util.find_spec(module_name)

    if spec is not None:
        try:
            version = metadata.version(package)
        except:
            try:
                version = metadata.version(module_name)
            except:
                version = "unknown"

        print(
            f"{package:<18} "
            f"{'OK':<12} "
            f"{version:<15} "
            f"{description}"
        )

    else:
        print(
            f"{package:<18} "
            f"{'THIẾU':<12} "
            f"{'-':<15} "
            f"{description}"
        )
        missing.append(package)

print("\n" + "=" * 120)
78
if missing:
    print("\nCác thư viện còn thiếu:")
    for lib in missing:
        print(f" - {lib}")

    print("\nLệnh cài:")
    print("pip install " + " ".join(missing))
else:
    print("\n Tất cả thư viện cần thiết đã được cài đặt.")
```
